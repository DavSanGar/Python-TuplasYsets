5. Crea un script que dada una tupla de números devuelva el producto de todos los números pares.

```python
tupla1 = (1,2,3,4,5,6,7,8)
producto = 1
for num in tupla1:
    if num % 2 == 0:
        producto = producto * num
print(producto)
```
